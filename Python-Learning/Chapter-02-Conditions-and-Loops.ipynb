# Chapter 2 - Conditions and Loops

Topics: if, elif, else, for loop, while loop, range(), break, continue.


## if / elif / else


```python
age = 25

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
marks = 72

if marks >= 75:
    print("Distinction")
elif marks >= 60:
    print("First Class")
else:
    print("Keep improving")
```

## for Loop and range()


```python
for i in range(1, 11):
    print(i)
```

```python
for i in range(2, 21, 2):
    print(i)
```

## while Loop


```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

## break and continue


```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```
