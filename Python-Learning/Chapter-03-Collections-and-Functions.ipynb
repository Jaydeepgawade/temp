# Chapter 3 - Collections and Functions

Topics: list, tuple, set, dictionary, functions, parameters, return values.


## List
A list stores multiple values and can be changed.


```python
skills = ["C#", "SQL", "Python"]
print(skills)
print(skills[0])
skills.append("AI")
print(skills)
```

## Tuple
A tuple is ordered but normally not changed after creation.


```python
coordinates = (18.52, 73.85)
print(coordinates)
print(coordinates[0])
```

## Set
A set keeps unique values.


```python
numbers = {1, 2, 2, 3, 3, 4}
print(numbers)
```

## Dictionary
Dictionary stores key-value pairs, similar to working with named properties.


```python
developer = {
    "name": "Jaydeep",
    "technology": ".NET",
    "experience": 3
}

print(developer)
print(developer["name"])
developer["learning"] = "Python"
print(developer)
```

## Functions


```python
def greet(name):
    print("Hello", name)

greet("Jaydeep")
```

## Function with return value


```python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

## Small Exercise - Employee Search


```python
employees = [
    {"id": 1, "name": "Amit"},
    {"id": 2, "name": "Jaydeep"},
    {"id": 3, "name": "Rahul"}
]

def find_employee(employee_id):
    for employee in employees:
        if employee["id"] == employee_id:
            return employee
    return None

print(find_employee(2))
```
