## Variables and Data Types
Python does not require explicit type declarations like C#.


```python
name = "Jaydeep"
age = 25
salary = 50000.50
is_developer = True

print(name)
print(age)
print(salary)
print(is_developer)
print(type(name), type(age), type(salary), type(is_developer))
```

## User Input and Type Conversion


```python
name = input("Enter your name: " )
age = int(input("Enter your age: "))
print("Hello", name)
print("Next year your age will be", age + 1)
```

## Operators


```python
a = 10
b = 3
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```
